# Selenium and Chromedriver

We use Selenium and ChromeDriver for collecting web data.

> Download the latest ChromeDriver from https://googlechromelabs.github.io/chrome-for-testing/

1. Define webdirver

```python
from selenium import webdriver

driver_path = './chromedriver.exe' # windows
# driver_path = './chromedriver'     # mac

cService = webdriver.ChromeService(executable_path=driver_path)
driver = webdriver.Chrome(service=cService)
driver.implicitly_wait(3) # waiting 3 seconds
```

2. Access the url

```python
url = 'https://sports.daum.net/worldsoccer/news/breaking'

# access the url
driver.get(url)
driver.implicitly_wait(2) # wait 2 seconds for loading the webpage
```

```python
# if you want to load more news, you can use `a.link_moreview` element in HTML
load_more = driver.find_element('css selector', 'a.link_moreview')
load_more.click()
```

3. Close the driver if your job is completed.

```python
driver.close()
```
