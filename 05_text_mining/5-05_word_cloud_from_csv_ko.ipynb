# Word Cloud using CSV (Ko)

Data: Movie Reviews (`./data/movie_reviews_ko.csv`)

### 1. Load Data

```python
import pandas as pd

df = pd.read_csv('./data/movie_reviews_ko.txt', sep='\t')
df.head()
```

### 2. Text Preprocessing

```python
from konlpy.tag import Okt

def get_words_ko(text, stopwords=None):
    okt = Okt()
    tagged = okt.pos(text, 
                     stem=True, 
                     norm=True)
    filtered = [w for w, pos in tagged if pos[0] in ['N', 'V', 'J']]  # Noun, Verb, Adjective
    filtered = [w for w in filtered if len(w) > 1]  # Remove single-character words
    
    if stopwords is not None:
        filtered = [w for w in filtered if w not in stopwords]
    return filtered
```

```python
custom_stopwords = ['영화', '보고', '보다', '하다', '이다', '않다', '으로', '에서']
df['words'] = df['document'].apply(lambda x: get_words_ko(x, stopwords=custom_stopwords))
```

### 3. Draw Wordclouds

#### 3-1. All reviews

```python
from collections import Counter
import matplotlib.pyplot as plt
from wordcloud import WordCloud

def draw_wordcloud(words, title=None):
    word_freq = Counter(words)
    wordcloud = WordCloud(font_path='./NanumSquareRoundR.ttf', 
                          width=800, height=400, background_color='white').generate_from_frequencies(word_freq)
    
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud)
    if title:
        plt.title(title)
    plt.axis('off')
    plt.show()
```

```python
import itertools

def flatten_list(list_of_lists):
    return list(itertools.chain.from_iterable(list_of_lists))
```

```python
words_all = flatten_list(df['words'].tolist())

draw_wordcloud(words_all, title='Word Cloud of All Reviews')
```

#### 3-2. Positive/Negative Reviews


```python
pos_reviews_words = df.loc[df['label']==1, 'words']
neg_reviews_words = df.loc[df['label']==0, 'words']

print('Number of positive reviews:', len(pos_reviews_words))
print('Number of negative reviews:', len(neg_reviews_words))
```

```python
words_pos = flatten_list(pos_reviews_words.tolist())
draw_wordcloud(words_pos, title='Word Cloud of Positive Reviews')

words_neg = flatten_list(neg_reviews_words.tolist())
draw_wordcloud(words_neg, title='Word Cloud of Negative Reviews')
```
