# Iris Data Exploration using  `seaborn`

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv('./data/Iris.csv')
df.head() # check if the dataframe is loaded properly
```

## 2. Data Visualization using `matplotlib`

##### Draw a histogram using matplotlib

```python
# import package
import matplotlib.pyplot as plt

# histogram
plt.hist(df['PetalWidthCm'], bins=10) # number of bins: 10

# details
plt.xlabel('PetalWidth')
plt.ylabel('Frequency')
plt.title('Histogram of PetalWidthCm')
```

# 3. Data Visualization using `seaborn`

### 3-1. Histogram

##### Draw a histogram using seaborn

```python
# import package
import seaborn as sns

# histogram
sns.histplot(df['PetalWidthCm'], bins=10).set(title='Histogram of PetalWidthCm')
```

```python
# import package

# histogram (hue added)
sns.histplot(x=df['PetalWidthCm'], bins=10, hue=df['Species']).set(title='Histogram of PetalWidthCm')
# sns.histplot(df, x='PetalWidthCm', bins=10, hue='Species').set(title='Histogram of PetalWidthCm') # the same
```

### 3-2. Box Plot

```python
df
```

```python
# for the box plot, we will use four numeric columns only
df_num = pd.concat([df.iloc[:, :4], df.iloc[:, -2:]], axis=1) # slicing syntax
df_num
```

```python
# boxpot
sns.boxplot(df_num).set(title='Boxplot of Iris')
```

### 3-3. Scatter Plot

```python
# scatter plot
sns.scatterplot(x=df['SepalLengthCm'], 
                y=df['PetalLengthCm']).set(title="Scatter Plot")
```

```python
# scatter plot (hue added)
sns.scatterplot(x=df['SepalLengthCm'], 
                y=df['PetalLengthCm'],
                hue=df['Species']).set(title="Scatter Plot")

# equivalent to the above
# sns.scatterplot(df, 
#                 x='SepalLengthCm', 
#                 y='PetalLengthCm',
#                 hue='Species').set(title="Scatter Plot")
```

### 3-4. Scatter matrix

```python
# scattermatrix
g = sns.pairplot(df, hue='Species')
plt.savefig('scattermatrix.png') # save the figure as a png file
```
