# Log File Analyzer

You are given a log file named `app.log`. Each line has a log level (`INFO`, `WARNING`, `ERROR`) followed by a colon and a message. 

Write a script that reads this file and counts how many logs exist for each level.

**Sample `app.log`** (located in `./data` directory):

```
INFO:Application has started.
WARNING:Disk space is running low.
INFO:User 'admin' logged in.
ERROR:Failed to connect to database.
INFO:Processing request #123.
WARNING:API response is slow.
ERROR:Null pointer exception at line 4
```

**Expected Output**:

```
Log Level Counts:
INFO: 3
WARNING: 2
ERROR: 3
```
