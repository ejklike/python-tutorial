# Classes

Imagine you're managing data for three different cars. 





## Without Classes:
Without classes, your code might look like this:

```python
# Car 1 data
car1_id = "red"
car1_speed = 0

# Car 2 data
car2_id = "blue"
car2_speed = 0

# Car 3 data
car3_id = "green"
car3_speed = 0

# Function to accelerate a car
def accelerate_car(car_id, current_speed, accel_level):
    new_speed = current_speed + accel_level * 10
    print(f"The {car_id} car is now going {new_speed} km/h.")
    return new_speed

# Using the function
car1_speed = accelerate_car(car1_id, car1_speed, 3)
car2_speed = accelerate_car(car2_id, car2_speed, 2)
```

## With Classes: 

Now, let's use a `Car` class to model the same problem.

- The `Car` class has two attributes: `id` and `speed`
- The `Car` class has two methods: `__init__()` and `accelerate()`
