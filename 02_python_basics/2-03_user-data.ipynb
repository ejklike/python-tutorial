# Website User Data

You have a list of user profiles, where each profile is a list containing the user's name and age.

**User Data:** `[['Alice', 25], ['Bob', 30], ['Charlie', 22], ['David', 35], ['Eve', 25]]`

**Problem:** Create a new list containing only the names of the users who are **30 years old or younger**.

**Expected Output:**

```
['Alice', 'Charlie', 'Eve', 'Bob']
```
