# 4. Data Transformation

### 4.1 Filter Rows

```python
# Filter Rows based on column values
filtered_df = df[df['PetalLengthCm'] > 6.0]
filtered_df
```

### 4.2 Add New Columns

```python
# Add New Columns
df['PetalArea'] = df['PetalLengthCm'] * df['PetalWidthCm']
df['Id'] = df.index
df['Source'] = 'UCI Repository'
```

### 4.3. Remove Columns

```python
# Remove single column
df = df.drop('Id', axis=1) # you must re-assign the variable to remove the column from the original df
# del df['Id'] # is equivalent to the line above

# remove multiple columns
df = df.drop(['PetalArea', 'Source'], axis=1) #
```

### 4.4 Apply a custom function using `apply()`

```python
# apply a function on a single column
def shorten_species(x):
    return x.split('-')[1] # 'Iris-setosa' splits into ['Iris', 'setosa']

df['ShortSpecies'] = df['Species'].apply(shorten_species)

df[['Species', 'ShortSpecies']].head()
```

```python
# apply with lambda on a single column
df['PetalSize'] = df['PetalLengthCm'].apply(
    lambda length: 'short' if length < 3.0 else 'long')

df[['PetalLengthCm', 'PetalSize']].head()
```

```python
# apply a custom function on a dataframe
def calculate_petal_ratio(x): # x is a row in a dataframe
    ratio = x['PetalLengthCm'] / x['PetalWidthCm']
    return ratio

# the function is applied to every row (axis=1)
df['PetalRatio'] = df.apply(calculate_petal_ratio, axis=1)

# the above is equivalent to the below
# df['PetalRatio'] = df.apply(
#     lambda row: row['PetalLengthCm'] / row['PetalWidthCm'], 
#     axis=1)

df[['PetalLengthCm', 'PetalWidthCm', 'PetalRatio']].head()
```
