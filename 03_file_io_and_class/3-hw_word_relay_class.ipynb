# Problem: Make a Word Relay Class

Create a Python class named `NextWordFinder`. 

**Requirements:**
1. The `__init__(self, txt_path)` method:
- The class should be initialized with a single argument: the text file path `txt_path` containing a list of words
- Upon initialization, the list of words should store the list as `self.words`.

2. The `find_next_word(self, word)` method:
- This method should accept one argument: a word (string)
- It should finds words in the `self.words` list that start with the last character (or syllable) of the input word.


Create two instances of the class `NextWordFinder`: 

- `next_word_finder_ko` using korean words (`./data/korean_words.txt`)
- `next_word_finder_en` using english words (`./data/english_words.txt`)
